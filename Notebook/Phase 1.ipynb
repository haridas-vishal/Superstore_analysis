# Superstore Sales Analysis: Phase 1 Notes

## About the dataset
This dataset contains order records of a retail superstore. It has 9,994 rows
and 21 columns. Each row is one product line within an order, with details
about the customer, location, product, and the sales and profit made.

## Column groups

**Order columns**
- Row ID: unique number for each row
- Order ID: ID of the order
- Order Date: date the order was placed
- Ship Date: date the order was shipped
- Ship Mode: shipping method (e.g. Standard Class,First class, Same day, Second class)

**Customer columns**
- Customer ID: unique ID of the customer
- Customer Name: name of the customer
- Segment: type of customer (Consumer, Corporate, Home Office)

**Location columns**
- Country, State, City, Postal Code: where the order was delivered
- Region: area of the country (West, East, Central, South)

**Product columns**
- Product ID: unique ID of the product
- Category: main product type (Furniture, Office Supplies, Technology)
- Sub-Category: more specific type (e.g. Chairs, Phones)
- Product Name: full name of the product

**Money columns**
- Sales: sales amount of the line
- Quantity: number of units sold
- Discount: discount given (0 to 0.8, i.e. 0% to 80%)
- Profit: profit or loss on the line

## Data types
- **Text:** Order ID, Ship Mode, Customer ID, Customer Name, Segment, Country,
  City, State, Region, Product ID, Category, Sub-Category, Product Name
- **Numbers:** Row ID, Postal Code, Sales, Quantity, Discount, Profit
- **Dates:** Order Date, Ship Date (currently stored as text, to be
  converted in Phase 2)

## Observations from describe()
- Profit ranges from about -6,600 (a loss) to about 8,400.
- Negative profit means the order lost money.
- Most orders are profitable (25th percentile of profit is positive).
- Average discount is about 15.6%, with a maximum of 80%.
- Row ID and Postal Code are identifiers, so their averages are not meaningful.
