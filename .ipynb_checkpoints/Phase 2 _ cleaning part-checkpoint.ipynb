```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
raw = pd.read_csv("Sample - Superstore.csv", encoding="latin-1")
df = raw.copy()
```

raw is the original. I’ll clean df. If I make a mistake, rerun this cell to start fresh.
While cleaning, you will rename columns, convert dates, and add new columns. If you make a mistake, you can compare df against raw to see what changed, or restart from raw with:

```python
df.columns = (df.columns.str.strip()
                        .str.lower()
                        .str.replace(" ", "_")
                        .str.replace("-","_")
             )
```

#### .str.strip()	Removes extra spaces at the start or end	(" Region" → "Region")
#### .str.lower()	Makes everything lowercase	(Order Date → order date)
#### .str.replace(" ", "_")	Replaces spaces with underscores	(order date → order_date)
#### .str.replace("-", "_")	Replaces hyphens with underscores	(sub-category → sub_category)

### Checked for missing values: none found in any of the 21 columns.” A check that finds nothing is still part of the cleaning process.

### check for Duplicate Values

```python
print("Duplicate rows:", df.duplicated().sum())
```

### Note : 0 means no duplicates, so there is nothing to remove. That’s a good result.

##### In this dataset, every row has a unique row_id, so no row can be an exact copy of another. That’s why the result is normally 0. 
###### Because row_id is always different, a repeated record could hide behind it. This checks every column except row_id:
###### print("Duplicates ignoring row_id:", df.drop(columns="row_id").duplicated().sum())

```python
print("Duplicates ignoring row_id:", df.drop(columns="row_id").duplicated().sum())
```

```python
dupes = df[df.drop(columns="row_id").duplicated(keep=False)]
dupes
```

```python
df = df.drop_duplicates(subset=df.columns.drop("row_id"))
print(df.shape)
```

#### subset=... means “compare rows using all columns except row_id”
#### By default it keeps the first copy and removes the later one

```python
print("Duplicates ignoring row_id:", df.drop(columns="row_id").duplicated().sum())
```

### Convert the Date Columns

```python
df[["order_date", "ship_date"]].head()
```

```python
# check data type
# datatype = Object(Text)
df[["order_date", "ship_date"]].dtypes
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], format="%m/%d/%Y")
df["ship_date"] = pd.to_datetime(df["ship_date"], format="%m/%d/%Y")
```

```python
# errors 
# for postal code no need of mean, max, min average etc
df["postal_code"].min()
# the postal code is "01040" but when we do min func the first digit "0" is not counted, this column is not for -
# - numerical operations
```

```python
df["postal_code"] = df["postal_code"].astype(str).str.zfill(5)
```

### Clean the Text Columns

```python
text_cols = df.select_dtypes(include="object").columns
print(text_cols.tolist())

for c in text_cols:
    df[c] = df[c].str.strip()
```

```python
# check for same names in columns(inconsistent spellings or "west"/"WEST" )
for c in ["ship_mode", "segment", "region", "category"]:
    print(c, "->", df[c].unique())
```

```python
df["region"].value_counts()
```

#### Check for Wrong or Impossible Values

```python
print("Ship date before order date:", (df["ship_date"] < df["order_date"]).sum())
print("Sales <= 0:", (df["sales"] <= 0).sum())
print("Quantity <= 0:", (df["quantity"] <= 0).sum())
print("Discount outside 0-1:", ((df["discount"] < 0) | (df["discount"] > 1)).sum())
```

```python
df[df["ship_date"] < df["order_date"]]
```

```python
print("Ship date before order date:", (df["ship_date"] < df["order_date"]).sum())
print("Sales <= 0:", (df["sales"] <= 0).sum())
print("Quantity <= 0:", (df["quantity"] <= 0).sum())
print("Discount outside 0-1:", ((df["discount"] < 0) | (df["discount"] > 1)).sum())
```

#### Create New Useful Columns

```python
df["order_year"] = df["order_date"].dt.year
df["order_month"] = df["order_date"].dt.month
df["order_month_name"] = df["order_date"].dt.month_name()
df["shipping_days"] = (df["ship_date"] - df["order_date"]).dt.days
df["profit_margin"] = (df["profit"] / df["sales"]).round(4)
df["is_loss"] = df["profit"] < 0
```
